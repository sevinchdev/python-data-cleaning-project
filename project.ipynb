STEP 2: Simple string cleaning

```python
# Removing scpace in each string column using strip()
df = df.apply(lambda col: col.str.strip() if col.dtype == 'object' else col)
df
```

```python
import numpy as np
```

```python
# Replacing empty strings with NaN
df = df.replace("", np.nan)
```

```python
df.isna().sum()
```

STEP 3: Cleaning numbers and dates

```python
numeric_columns = ['age', 'score','gpa','attendance', 'money_spent']

for col in numeric_columns:

  # Cleaning from symbols
  df[col] = df[col].astype(str).str.replace(r"[^0-9.-]", "", regex=True)

  # Convert to numeric, invalid values bacome NaN
  df[col] = pd.to_numeric(df[col], errors='coerce')

df.loc[df['age']<0, 'age'] = np.nan
df.loc[(df['gpa']<0) | (df['gpa']>5), 'gpa'] = np.nan
df.loc[(df['attendance']<0) | (df['attendance']>100), 'attendance'] = np.nan
df.loc[(df['score']<0) | (df['score']>100), 'score'] = np.nan
df.loc[df['money_spent']<0, 'money_spent'] = np.nan

df['age'] = df['age'].astype('Int64')
df['gpa'] = df['gpa'].astype(float)
df['attendance'] = df['attendance'].astype(float)
df['score'] = df['score'].astype(float)
df['money_spent'] = df['money_spent'].astype(float)
```

```python
df
```

```python
df.isna().sum()
```

```python
# Converting date columns into pandas datetime format
date_columns = ['date_of_join', 'event_time']

for col in date_columns:
  df[col] = pd.to_datetime(df[col], errors='coerce')

  df[col] = df[col].dt.strftime("%Y-%m-%d %H:%M:%S")
```

```python
df
```

STEP 4: Email and Phone validation

```python
# Making sure that empty or NaN stays as NaN
df['email'] = df['email'].replace(r'^\s*$', np.nan, regex=True)

# Stripping spaces
df['email'] = df['email'].str.strip()

# Turning emails into lowercase 
df['email'] = df['email'].str.lower()

# Regex for valid email
email_pattern = r'^[a-z0-9._%+-]+@[a-z0-9.-]+\.[a-z]{2,}$'

valid_mask = df['email'].str.match(email_pattern, na=False)

df.loc[~valid_mask, 'email'] = np.nan
```

```python
df.head(20)
```

```python
# cleaning phone numbers and converting them into standart form
df['phone'] = df['phone'].astype(str)

df['phone'] = df['phone'].str.replace(r'(ext|x|extension)\s*\d+', "", regex = True)
df['phone_clean'] = df['phone'].str.replace(r'\D',"", regex = True)

df['phone_clean'] = "+" + df['phone_clean']

# regex for phone number
phone_pattern =  r'^\+\d{8,15}$'

valid_mask = df['phone_clean'].str.len().between(9, 16)
df.loc[~valid_mask, 'phone_clean'] = np.nan

df['phone'] = df['phone_clean']
df.drop(columns=['phone_clean'], inplace= True)
```

```python
df.head(20)
```

STEP 5: JSON Parsing

```python
import json
```

```python
# Conversting Json into Python dictionary

df['profile_json'] = df['profile_json'].astype(str).str.replace("'",'"', regex = False)

def json_parse(x):
  try:
    return json.loads(x)
  except:
    return {}
  
df['profile_json'] = df['profile_json'].apply(json_parse)
```

```python
df
```

```python
# Flatten Json into columns

json_df = pd.json_normalize(df['profile_json'])
```

```python
df = pd.concat([df.drop(columns=['profile_json']), json_df], axis = 1)
```

```python
df.head(20)
```

```python
def flatten_devices(devices_list):
  result = {}

  if isinstance(devices_list, list):
    for i, device in enumerate(devices_list):
      for key, value in device.items():
        result[f"device{i+1}_{key}"] = value

  return result
```

```python
names = ['alex','brown','tom']

for i, name in enumerate(names):
  print(f"{i} {name}")
```

```python
devices_expanded = df['devices'].apply(flatten_devices).apply(pd.Series)
```

```python
df = pd.concat([df.drop(columns = ['devices']), devices_expanded], axis=1)
```

STEP 8: Data Normalization

```python
def normalize_gender(x):
  x = str(x).strip().lower()
  if x in ['male','m']:
    return 'Male'
  elif x in ['female','f','fmale']:
    return 'Female'
  else:
    return 'Unknown'
  
df['gender'] = df['gender'].apply(normalize_gender)
```

```python
def normalize_course(x):
  x = str(x).strip().lower()
  if 'data' in x and 'science' in x:
    return 'Data Science'
  elif 'python' in x:
    return 'Python'
  else:
    return 'Unknown'
  
df['course'] = df['course'].apply(normalize_course)
```

```python
df['status'] = df['status'].astype(str).str.strip().str.lower()
```
